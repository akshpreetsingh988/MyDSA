### Frequency of elements 

```python
def solve(arr): 
    mapp ={} 
    ans = [] 
    for i in range(len(arr)) : 
        mapp[arr[i]]= mapp.get(arr[i], 0) +1
    for x,y in mapp.items(): 
        ans.append([x,y] )
    return ans 
    
if __name__ == "__main__" :
    arr= [10, 20, 10,5] 
    ans = solve(arr) 
    print(f"ans : {ans}")
```
