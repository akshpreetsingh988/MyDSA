### Frequency of elements 

```python
def solve(arr): 
    mapp ={} 
    ans = [] 
    for i in range(len(arr)) : 
        mapp[arr[i]]= mapp.get(arr[i], 0) +1
    for x,y in mapp.items(): 
        ans.append([x,y] )
    return ans 
    
if __name__ == "__main__" :
    arr= [10, 20, 10,5] 
    ans = solve(arr) 
    print(f"ans : {ans}")
```

### Hashing

```python
# implementing quadratic hashing 

def quadratic_hashing(arr , m) : 
    hashTable = [-1] * m 
    for x in arr: 
        index = x%m 
        i=0 
        if hashTable[index] ==-1: 
            hashTable[index]=x 
        else: 
            i=0 
            while True: 
                pos = (index + i*i) %m
                if hashTable[pos]==-1: 
                    hashTable[pos] = x
                    break  
                elif hashTable[pos]==x: 
                    break 
                i+=1 
                if i==m: 
                    break 
        
    print(hashTable) 
if __name__== "__main__" : 
    arr=  [31 , 48 , 37, 70] 

    hashSize = 4
    quadratic_hashing(arr, hashSize) 
```

### Anagram

```python
def isAnagram(s1, s2) : 
    mapp1 = {}
    mapp2 = {} 
    for x in s1: 
        mapp1[x] = 1+ mapp1.get(x,0)
    for x in s2: 
        mapp2[x] = 1+ mapp2.get(x,0)
    return  mapp1 == mapp2
s1 = "geeks" 
s2 = "kseeg" 
print(isAnagram(s1, s2) )
```

### target sum 

```python
# 2 sum 
def pairSum(arr, target) : # using a sett can also be done using sorting and keeping left and right pointers 
    # pass 
    sett = set() 
    for i in range(len(arr)) :
        x= target - arr[i]
        if x in sett: 
            return True 
        sett.add(arr[i]) 
    return False 
    
arr=[0, -1, -3, 1,2 ]  
target = -2
print(pairSum(arr , target) 
)
```

### Union Find of 2 arrays

```python
def unionFind(arr1, arr2) : 
    # pass
    sett = set() 
    for x in arr1: 
        sett.add(x) 
    for x in arr2: 
        sett.add(x) 
    return list(sett) 

if __name__ == "__main__": 
    arr1 = [1,2,3,2,1] 
    arr2 = [3,2,2,3,3,2] 

    ans= unionFind(arr1, arr2) 
    print(ans)
```

### Intersection of 2 arrays 

```python
def intersect(a, b) : 
    ans = [] 
    sett = set() 
    for x in a: sett.add(x) 
    for x in b : 
        if x in sett: 
            ans.append(x) 
    return list(set(ans))

intersect([1,2,1,3,1] , [3,1,3,4,1] ) 
```

### Isomorphic Strings

```python
# check the mapping of s1 to s2 
# isomorphic strings 

def solve(s1, s2) : 
    # pass 
    mapp1= {} 
    mapp2 = {}
    for c1, c2 in zip(s1, s2) : 
        if c1 in mapp1  and mapp1[c1] != c2: 
            return False 
        if c2 in mapp2  and mapp2[c2] != c1: 
            return False 
        mapp1[c1] = c2 
        mapp2[c2]= c1 
    return True 

if __name__ == "__main__": 
    s1= "aab" 
    s2= "xxy"
    ans = solve(s1, s2)  
    print(ans ) 
```

### prefx sum range query 

```python
# [prefx] sum range query 
def solve(arr , queries ) : 
    # pass 
    n= len(arr)
    res=[]
    prefixSum = [0]*n
    prefixSum[0] = arr[0] 
    for i in range(1, n) : 
        prefixSum[i] = prefixSum[i-1]+arr[i]
    print(prefixSum) 
    for l,r in queries:
        if l==0:
            ans= prefixSum[r] 
        if l>0: 
            ans =prefixSum[r]-prefixSum[l-1]
        res.append(ans)
    print(f"res : {res}")
```
