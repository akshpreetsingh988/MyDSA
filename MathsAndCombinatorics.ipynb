```python
# finding primes using sieve of eratosthenes 
if __name__ == "__main__":
    n = 100
    primes = [True] * (n + 1)

    primes[0] = False
    primes[1] = False

    p = 2
    while p * p <= n:
        if primes[p] == True:
            for i in range(p * p, n + 1, p):
                primes[i] = False
        p += 1

    ans = []
    print(primes)
    for i in range(len(primes)):
        if primes[i] == True:
            ans.append(i)
    print(ans)

```

```python
# check prime optimized way 
def check_prime(n):
    if n == 0 or n == 1:
        return False
    if n == 2 or n == 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i <= (n**0.5):
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True


if __name__ == "__main__":
    ans = check_prime(13)
    print(ans)

```

```python
# binary exponentiation  to find a power b mod M
def binary_expo(a, b, MOD):
    res = 1

    while b >= 1:
        print(f"b : {b} , a : {a}, res : {res}")
        if b % 2 == 1:  # odd
            res = (res * a) % MOD
        # else:  # even
        a = (a * a) % MOD
        b = b // 2
    return res


if __name__ == "__main__":
    a = 2
    b = 3
    MOD = (10**9) + 7
    ans = binary_expo(a, b, MOD)
    print(f"ans : {ans}")
```

```python
# finding number closest to n divisible my m 
def solve(n, m) : 
    quotient = n//m 
    lower = quotient*m 
    upper= (quotient+1)*m 

    abs_disfromlower = abs(n - lower )
    abs_disfromupper= abs(n - upper)

    return lower if abs_disfromlower<abs_disfromupper else upper
solve(13, 4) 
```

```python
# finding square root of a number using binary search 
n =100 
def solve(n) : 
    l =1 
    r= n 
    ans =None 
    while l<=r: 
        mid = (l+r)//2 
        if mid*mid <= n: 
            ans = mid 
            l=mid+1
        else: 
            r =mid-1 
    return ans 
solve(n ) 
```

```python
# distinct divisors 
def solve(n ): 
    ans=[] 
    for i in range(1 , int(n**0.5)+1): 
        if n%i==0: 
            if n//i==i : 
                ans.append(i)
            else: 
                ans.append(i) 
                ans.append(n//i) 
    return ans 
print(solve(100))
```

```python
# calculating nCr 
def factorial(n) : 
    ans= 1
    for i in range(2, n+1):
        ans = ans*i 
    return ans  

def nCr(n, r) : 
    return factorial(n) // factorial(n-r) * factorial(r)

nCr(4 ,2)
```

```python
# power function
def power(a, b , mod=10**9+7):  
    res =1 
    while b>=1: 
        print(f"a: {a}, b: {b} ,res : {res}") 
        if b%2==1: 
            res = (res*a)%mod 
            b-=1 
        else: 
            a= (a*a)%mod 
            b = b//2 
    return res 
print(power(2, 6,))
```
