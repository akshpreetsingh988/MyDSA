```python
# finding primes using sieve of eratosthenes 
if __name__ == "__main__":
    n = 100
    primes = [True] * (n + 1)

    primes[0] = False
    primes[1] = False

    p = 2
    while p * p <= n:
        if primes[p] == True:
            for i in range(p * p, n + 1, p):
                primes[i] = False
        p += 1

    ans = []
    print(primes)
    for i in range(len(primes)):
        if primes[i] == True:
            ans.append(i)
    print(ans)

```

```python
# check prime optimized way 
def check_prime(n):
    if n == 0 or n == 1:
        return False
    if n == 2 or n == 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i <= (n**0.5):
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True


if __name__ == "__main__":
    ans = check_prime(13)
    print(ans)

```

```python
# binary exponentiation  to find a power b mod M
def binary_expo(a, b, MOD):
    res = 1

    while b >= 1:
        print(f"b : {b} , a : {a}, res : {res}")
        if b % 2 == 1:  # odd
            res = (res * a) % MOD
        # else:  # even
        a = (a * a) % MOD
        b = b // 2
    return res


if __name__ == "__main__":
    a = 2
    b = 3
    MOD = (10**9) + 7
    ans = binary_expo(a, b, MOD)
    print(f"ans : {ans}")
```

```python
# finding number closest to n divisible my m 
def solve(n, m) : 
    quotient = n//m 
    lower = quotient*m 
    upper= (quotient+1)*m 

    abs_disfromlower = abs(n - lower )
    abs_disfromupper= abs(n - upper)

    return lower if abs_disfromlower<abs_disfromupper else upper
solve(13, 4) 
```
